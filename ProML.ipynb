```python
from ultralytics import YOLO
model = YOLO("yolov8n.pt")
```

```python
model.train(
    data="dataset.yaml",
    epochs=30,
    imgsz=640,
    batch=16,
    name="weapon_detection"
)
```

```python
from IPython.display import Image
Image("runs/detect/weapon_detection/results.png")
```

```python
model = YOLO("runs/detect/weapon_detection/weights/best.pt")

results = model("test.jpeg", conf=0.4, save=True)
results[0].show()
```

```python
for r in results:
    for box in r.boxes:
        conf = float(box.conf[0])

        if conf > 0.7:
            print("🔴 HIGH THREAT")
        elif conf > 0.4:
            print("🟡 MEDIUM THREAT")
        else:
            print("🟢 LOW THREAT")
```
